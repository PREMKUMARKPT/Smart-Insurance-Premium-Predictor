```python

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import matplotlib.pyplot as plt
import pickle
```

```python
import pandas as pd
```

```python
df = pd.read_csv("/content/train - Copy.csv")
```

```python
# handling the missing values

for i in df.select_dtypes(include='object').columns:
  df[i].fillna(df[i].mode()[0], inplace=True)
```

```python
for i in df.select_dtypes(include=["int64","float64"]).columns:
  df[i] = df[i].fillna(df[i].mean())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.head()
```

```python
df["Premium Amount"] = np.log1p(df["Premium Amount"])
```

```python
x = df.drop("Premium Amount", axis=1)
y = df["Premium Amount"]
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42, test_size=0.2)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
x_test.shape
```

```python
y_test.shape
```

```python
# Identify numeric and categorical columns
numeric_cols = x.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = x.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
numeric_cols
```

```python
categorical_cols
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
# Preprocessor
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), numeric_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
])

# Pipeline
model = Pipeline([
    ('preprocess', preprocessor),
    ('regressor', RandomForestRegressor(random_state=42, n_jobs=-1))
])

# Fit model
model.fit(x_train, y_train)
```

```python
# Evaluate
print("🔹 Train R²:", r2_score(y_train, model.predict(x_train)))
print("🔸 Test R²:", r2_score(y_test, model.predict(x_test)))
```

```python
# Save model + features together
save_object = {
    'model': model,
    'features': x.columns.tolist()
}
with open('model_and_features.pkl', 'wb') as f:
    pickle.dump(save_object, f)

print("✅ Model and feature list saved to model_and_features.pkl")
```
